#Gabung Headline dan Isi Berita


```python
# FUNGSI GABUNG HEADLINE + 5 KALIMAT
def gabung_headline_isi(row):
    headline = str(row['Headline']).strip()
    isi_mentah = str(row['Isi Berita'])

    list_kalimat = isi_mentah.split('.')

    kalimat_valid = [k.strip() for k in list_kalimat if len(k.strip()) > 20]


    lima_kalimat = '. '.join(kalimat_valid[:5])

    return headline + ". " + lima_kalimat + "."
#buat kolom baru
df['teks_gabung'] = df.apply(gabung_headline_isi, axis=1)
```

#Text Cleaning for Data ML

#Teks CLeaning

```python
import re
import string

emoji_pattern = re.compile("["        
    u"\U0001F600-\U0001F64F"
    u"\U0001F300-\U0001F5FF"
    u"\U0001F680-\U0001F6FF"
    u"\U0001F1E0-\U0001F1FF"
    u"\U00002702-\U000027B0"
    u"\U000024C2-\U0001F251"
    "]+", flags=re.UNICODE)

def cleaningText(text):
    if not isinstance(text, str):
        return ''                                

    text = re.sub(r'@[A-Za-z0-9]+', '', text)     # hapus mention
    text = re.sub(r'#[A-Za-z0-9]+', '', text)     # hapus hashtag
    text = re.sub(r'RT[\s]', '', text)            # hapus retweet
    text = re.sub(r'http\S+', '', text)           # hapus URL
    text = emoji_pattern.sub(r'', text)           
    text = text.replace('\n', ' ')              
    text = re.sub(r'[^A-Za-z ]+', '', text)       # hapus non-alfabet (angka, punctuation, sisa emoji)
    text = re.sub(r'\s+', ' ', text).strip()      

    return text

df['data_clean'] = df['teks_gabung'].apply(cleaningText)
df.head()
```

#casefolding

```python
#Lowercase
df['data_clean'] = df['data_clean'].apply(lambda x: str(x).lower() if pd.notnull(x) else '')
df.head()
```

#Stopword & Stemming

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import nltk
from nltk.corpus import stopwords

# Download data stopword dari NLTK
nltk.download('stopwords')
nltk.download('punkt')

# Inisialisasi Stemmer Sastrawi
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Load Stopwords Bahasa Indonesia
list_stopwords = set(stopwords.words('indonesian'))
```

```python
# 1. Load Stopwords & Custom Noise
list_stopwords = set(stopwords.words('indonesian'))
additional_noise = {'scroll', 'to', 'continue', 'with', 'content', 'detik'}
list_stopwords.update(additional_noise)

# 2. Eksekusi Stopword
print("Sedang menghapus stopwords...")
df['data_stopword'] = df['data_clean'].apply(lambda x: " ".join([t for t in x.split() if t.lower() not in list_stopwords]))

print("sesesai!")
```

```python
# 3. Eksekusi Stemming
print("Sedang memproses Stemming... Mohon tunggu.")
df['data_lemma'] = df['data_stopword'].apply(lambda x: stemmer.stem(x))

print('selesai!')

df.head()
```

#Tokenization & detokenization

```python
# SEL 4: Tokenizing & Detokenizing
df['data_token'] = df['data_lemma'].apply(lambda x: x.split())
```

```python
# Detokenize
df['detokenization'] = df['data_token'].apply(lambda x: ' '.join(x) if isinstance(x, list) else x)
```
